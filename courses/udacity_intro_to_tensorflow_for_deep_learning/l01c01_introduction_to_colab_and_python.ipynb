```
# ============================================
# 1. Imports
# ============================================
import torch
import torchvision
import torchvision.transforms as transforms
from torch.utils.data import DataLoader
import matplotlib.pyplot as plt
import numpy as np

# ============================================
# 2. Check GPU availability
# ============================================
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(f"Using device: {device}")

# ============================================
# 3. Load Fashion MNIST dataset
# ============================================
transform = transforms.Compose([
    transforms.ToTensor()   # converts PIL image (0-255) -> tensor (0.0-1.0), shape (1, 28, 28)
])

train_set = torchvision.datasets.FashionMNIST(
    root='./data', train=True, download=True, transform=transform
)
test_set = torchvision.datasets.FashionMNIST(
    root='./data', train=False, download=True, transform=transform
)

# Class labels for Fashion MNIST (0-9)
class_names = ['T-shirt/top', 'Trouser', 'Pullover', 'Dress', 'Coat',
               'Sandal', 'Shirt', 'Sneaker', 'Bag', 'Ankle boot']

# ============================================
# 4. Create DataLoaders (batches the data for training/viewing)
# ============================================
train_loader = DataLoader(train_set, batch_size=128, shuffle=True)
test_loader = DataLoader(test_set, batch_size=128, shuffle=False)

# ============================================
# 5. show_grid function
# ============================================
def show_grid(images, nrow=8, title=None, figsize=(10, 10), normalize=True):
    """
    Display a batch of images as a grid.

    Args:
        images (Tensor): batch of images, shape (B, C, H, W)
        nrow (int): number of images per row in the grid
        title (str): optional title for the plot
        figsize (tuple): size of the matplotlib figure
        normalize (bool): whether to normalize pixel values to [0, 1] for display
    """
    images = images.detach().cpu()

    # Arrange all images in the batch into a single grid image
    grid = torchvision.utils.make_grid(images, nrow=nrow, normalize=normalize, padding=2)

    # matplotlib expects (H, W, C), but tensors are (C, H, W) -> transpose
    np_grid = grid.numpy().transpose((1, 2, 0))

    plt.figure(figsize=figsize)
    if np_grid.shape[2] == 1 or (np_grid[:, :, 0] == np_grid[:, :, 1]).all():
        plt.imshow(np_grid[:, :, 0], cmap='gray')   # grayscale display
    else:
        plt.imshow(np_grid)

    plt.axis('off')
    if title:
        plt.title(title)
    plt.show()

# ============================================
# 6. Try it out — grab one batch and display it
# ============================================
images, labels = next(iter(train_loader))
print(f"Batch shape: {images.shape}")   # [64, 1, 28, 28]

show_grid(images, nrow=8, title="Sample Fashion MNIST Batch")

# Optional: print the labels for that batch
print("Labels:", [class_names[l] for l in labels[:8]])
```

```
# ============================================
# 2.1 Build the Model — Encoder + Decoder
# ============================================
import torch.nn as nn
import torch.optim as optim

LATENT_DIM = 32

class Encoder(nn.Module):
    def __init__(self, latent_dim=LATENT_DIM):
        super().__init__()
        self.net = nn.Sequential(
            nn.Conv2d(1, 32, kernel_size=3, stride=2, padding=1),   # 28x28 -> 14x14
            nn.ReLU(True),
            nn.Conv2d(32, 64, kernel_size=3, stride=2, padding=1),  # 14x14 -> 7x7
            nn.ReLU(True),
            nn.Flatten(),
            nn.Linear(64 * 7 * 7, latent_dim)
        )

    def forward(self, x):
        return self.net(x)


class Decoder(nn.Module):
    def __init__(self, latent_dim=LATENT_DIM):
        super().__init__()
        self.fc = nn.Linear(latent_dim, 64 * 7 * 7)
        self.net = nn.Sequential(
            nn.ReLU(True),
            nn.Unflatten(1, (64, 7, 7)),
            nn.ConvTranspose2d(64, 32, kernel_size=4, stride=2, padding=1),  # 7x7 -> 14x14
            nn.ReLU(True),
            nn.ConvTranspose2d(32, 1, kernel_size=4, stride=2, padding=1),   # 14x14 -> 28x28
            nn.Sigmoid()   # output in [0,1]
        )

    def forward(self, z):
        x = self.fc(z)
        return self.net(x)


class DenoisingAutoencoder(nn.Module):
    def __init__(self, latent_dim=LATENT_DIM):
        super().__init__()
        self.encoder = Encoder(latent_dim)
        self.decoder = Decoder(latent_dim)

    def forward(self, x):
        z = self.encoder(x)
        return self.decoder(z)


model = DenoisingAutoencoder(LATENT_DIM).to(device)
print(model)


# ============================================
# 2.2 Add Noise function
# ============================================
def add_noise(images, sigma=0.3):
    """
    Add Gaussian noise to a batch of images and clamp to valid [0,1] range.
    """
    noise = torch.randn_like(images) * sigma
    noisy = images + noise
    noisy = torch.clamp(noisy, 0.0, 1.0)
    return noisy


# ============================================
# 2.3 Train the Model
# ============================================
criterion = nn.MSELoss()
optimizer = optim.Adam(model.parameters(), lr=1e-3)

NUM_EPOCHS = 10

train_losses = []

for epoch in range(NUM_EPOCHS):
    model.train()
    running_loss = 0.0

    for images, _ in train_loader:
        images = images.to(device)                # clean images (target)
        noisy_images = add_noise(images, sigma=0.3)  # corrupted input

        optimizer.zero_grad()
        outputs = model(noisy_images)              # reconstruct from noisy input
        loss = criterion(outputs, images)          # compare to clean original

        loss.backward()
        optimizer.step()

        running_loss += loss.item() * images.size(0)

    epoch_loss = running_loss / len(train_loader.dataset)
    train_losses.append(epoch_loss)
    print(f"Epoch [{epoch+1}/{NUM_EPOCHS}] - MSE Loss: {epoch_loss:.6f}")

# ============================================
# Plot training loss
# ============================================
plt.figure(figsize=(6, 4))
plt.plot(range(1, NUM_EPOCHS + 1), train_losses, marker='o')
plt.xlabel("Epoch")
plt.ylabel("MSE Loss")
plt.title("Denoising Autoencoder Training Loss")
plt.grid(True)
plt.show()

# ============================================
# Visualize results: clean vs noisy vs reconstructed
# ============================================
model.eval()
with torch.no_grad():
    test_images, _ = next(iter(test_loader))
    test_images = test_images.to(device)
    noisy_test = add_noise(test_images, sigma=0.3)
    reconstructed = model(noisy_test)

show_grid(test_images[:8], nrow=8, title="Original (Clean)")
show_grid(noisy_test[:8], nrow=8, title="Noisy Input")
show_grid(reconstructed[:8], nrow=8, title="Reconstructed Output")
```

```

# ============================================
# 3.1 VAE Encoder Outputs — μ and logσ²
# ============================================
class VAEEncoder(nn.Module):
    def __init__(self, latent_dim=LATENT_DIM):
        super().__init__()
        self.conv = nn.Sequential(
            nn.Conv2d(1, 32, kernel_size=3, stride=2, padding=1),   # 28x28 -> 14x14
            nn.ReLU(True),
            nn.Conv2d(32, 64, kernel_size=3, stride=2, padding=1),  # 14x14 -> 7x7
            nn.ReLU(True),
            nn.Flatten()
        )
        self.fc_mu = nn.Linear(64 * 7 * 7, latent_dim)
        self.fc_logvar = nn.Linear(64 * 7 * 7, latent_dim)

    def forward(self, x):
        h = self.conv(x)
        mu = self.fc_mu(h)
        logvar = self.fc_logvar(h)
        return mu, logvar


# Decoder is the same architecture as before (reuse it)
class VAEDecoder(nn.Module):
    def __init__(self, latent_dim=LATENT_DIM):
        super().__init__()
        self.fc = nn.Linear(latent_dim, 64 * 7 * 7)
        self.net = nn.Sequential(
            nn.ReLU(True),
            nn.Unflatten(1, (64, 7, 7)),
            nn.ConvTranspose2d(64, 32, kernel_size=4, stride=2, padding=1),  # 7x7 -> 14x14
            nn.ReLU(True),
            nn.ConvTranspose2d(32, 1, kernel_size=4, stride=2, padding=1),   # 14x14 -> 28x28
            nn.Sigmoid()
        )

    def forward(self, z):
        x = self.fc(z)
        return self.net(x)


# ============================================
# 3.2 Reparameterisation trick
# ============================================
def reparameterize(mu, logvar):
    """
    z = mu + exp(0.5 * logvar) * epsilon,  epsilon ~ N(0, I)
    Lets gradients flow through mu and logvar despite the sampling step.
    """
    std = torch.exp(0.5 * logvar)
    eps = torch.randn_like(std)
    z = mu + std * eps
    return z


class VAE(nn.Module):
    def __init__(self, latent_dim=LATENT_DIM):
        super().__init__()
        self.encoder = VAEEncoder(latent_dim)
        self.decoder = VAEDecoder(latent_dim)

    def forward(self, x):
        mu, logvar = self.encoder(x)
        z = reparameterize(mu, logvar)
        x_recon = self.decoder(z)
        return x_recon, mu, logvar


vae = VAE(LATENT_DIM).to(device)
print(vae)


# ============================================
# 3.3 ELBO Loss = Reconstruction (BCE) + KL divergence
# ============================================
def vae_loss(x_recon, x, mu, logvar):
    # Reconstruction term: binary cross-entropy, summed over pixels
    recon_loss = nn.functional.binary_cross_entropy(x_recon, x, reduction='sum')

    # KL divergence between N(mu, sigma^2) and N(0, I), summed over latent dims
    kl_loss = -0.5 * torch.sum(1 + logvar - mu.pow(2) - logvar.exp())

    # Average over batch size for a stable, comparable loss scale
    batch_size = x.size(0)
    return (recon_loss + kl_loss) / batch_size, recon_loss / batch_size, kl_loss / batch_size


# ============================================
# 3.3 Train the VAE
# ============================================
vae_optimizer = optim.Adam(vae.parameters(), lr=2e-3)
NUM_EPOCHS_VAE = 12

vae_train_losses = []

for epoch in range(NUM_EPOCHS_VAE):
    vae.train()
    running_loss, running_recon, running_kl = 0.0, 0.0, 0.0

    for images, _ in train_loader:
        images = images.to(device)

        vae_optimizer.zero_grad()
        x_recon, mu, logvar = vae(images)
        loss, recon_loss, kl_loss = vae_loss(x_recon, images, mu, logvar)

        loss.backward()
        vae_optimizer.step()

        running_loss += loss.item() * images.size(0)
        running_recon += recon_loss.item() * images.size(0)
        running_kl += kl_loss.item() * images.size(0)

    n = len(train_loader.dataset)
    vae_train_losses.append(running_loss / n)
    print(f"Epoch [{epoch+1}/{NUM_EPOCHS_VAE}] - ELBO: {running_loss/n:.2f} "
          f"| Recon: {running_recon/n:.2f} | KL: {running_kl/n:.2f}")

# Plot loss
plt.figure(figsize=(6, 4))
plt.plot(range(1, NUM_EPOCHS_VAE + 1), vae_train_losses, marker='o')
plt.xlabel("Epoch")
plt.ylabel("ELBO Loss")
plt.title("VAE Training Loss")
plt.grid(True)
plt.show()


# ============================================
# Figure 2: Random samples from the prior
# ============================================
vae.eval()
with torch.no_grad():
    z_random = torch.randn(64, LATENT_DIM).to(device)   # sample from N(0, I)
    generated = vae.decoder(z_random)

show_grid(generated, nrow=8, title="Figure 2: Random Samples from Prior")


# ============================================
# Figure 3: Interpolations between two points in latent space
# ============================================
with torch.no_grad():
    images_batch, _ = next(iter(test_loader))
    images_batch = images_batch.to(device)
    mu, logvar = vae.encoder(images_batch)

    z_start = mu[0]
    z_end = mu[1]

    steps = 10
    alphas = torch.linspace(0, 1, steps).to(device)
    z_interp = torch.stack([z_start * (1 - a) + z_end * a for a in alphas])
    interp_images = vae.decoder(z_interp)

show_grid(interp_images, nrow=steps, title="Figure 3: Latent Space Interpolation")


# ============================================
# Figure 4+: Latent traversals (vary one dimension at a time)
# ============================================
with torch.no_grad():
    base_z = mu[0].clone()   # start from a real encoded image
    n_dims_to_show = 6       # traverse first 6 latent dims (out of 32)
    steps = 8
    traversal_range = torch.linspace(-3, 3, steps).to(device)

    for dim in range(n_dims_to_show):
        z_batch = base_z.unsqueeze(0).repeat(steps, 1)
        z_batch[:, dim] = traversal_range
        traversal_imgs = vae.decoder(z_batch)
        show_grid(traversal_imgs, nrow=steps, title=f"Figure {4+dim}: Traversal of latent dim {dim}")
```

```
# ============================================
# 4.1 cGAN Architecture
# ============================================
NUM_CLASSES = 10
Z_DIM = 100
LABEL_EMBED_DIM = 10

class Generator(nn.Module):
    def __init__(self, z_dim=Z_DIM, num_classes=NUM_CLASSES, embed_dim=LABEL_EMBED_DIM):
        super().__init__()
        self.label_embed = nn.Embedding(num_classes, embed_dim)

        self.net = nn.Sequential(
            # input: (z_dim + embed_dim) -> project to 7x7x128
            nn.Linear(z_dim + embed_dim, 128 * 7 * 7),
            nn.ReLU(True),
            nn.Unflatten(1, (128, 7, 7)),

            nn.ConvTranspose2d(128, 64, kernel_size=4, stride=2, padding=1),  # 7x7 -> 14x14
            nn.BatchNorm2d(64),
            nn.ReLU(True),

            nn.ConvTranspose2d(64, 1, kernel_size=4, stride=2, padding=1),   # 14x14 -> 28x28
            nn.Tanh()   # output in [-1, 1]
        )

    def forward(self, z, labels):
        label_emb = self.label_embed(labels)          # (B, embed_dim)
        x = torch.cat([z, label_emb], dim=1)           # concatenate z with label embedding
        return self.net(x)


class Discriminator(nn.Module):
    def __init__(self, num_classes=NUM_CLASSES):
        super().__init__()
        # Embed label into a full 28x28 map so it can be concatenated as an extra channel
        self.label_embed = nn.Embedding(num_classes, 28 * 28)

        self.net = nn.Sequential(
            nn.Conv2d(2, 64, kernel_size=3, stride=2, padding=1),   # 2 channels: image + label map
            nn.LeakyReLU(0.2, inplace=True),

            nn.Conv2d(64, 128, kernel_size=3, stride=2, padding=1),
            nn.BatchNorm2d(128),
            nn.LeakyReLU(0.2, inplace=True),

            nn.Flatten(),
            nn.Linear(128 * 7 * 7, 1)   # raw logit output (BCEWithLogits expects this)
        )

    def forward(self, img, labels):
        label_map = self.label_embed(labels).view(-1, 1, 28, 28)  # (B, 1, 28, 28)
        x = torch.cat([img, label_map], dim=1)                     # (B, 2, 28, 28)
        return self.net(x)


generator = Generator().to(device)
discriminator = Discriminator().to(device)
print(generator)
print(discriminator)
```

```
# ============================================
# 4.2 Train the cGAN
# ============================================
criterion = nn.BCEWithLogitsLoss()

g_optimizer = optim.Adam(generator.parameters(), lr=2e-4, betas=(0.5, 0.999))
d_optimizer = optim.Adam(discriminator.parameters(), lr=2e-4, betas=(0.5, 0.999))

NUM_EPOCHS_GAN = 15
REAL_LABEL_SMOOTH = 0.9   # label smoothing for real samples (stability trick)

# Re-normalize data loader to [-1, 1] for Tanh output — build a fresh loader
transform_gan = transforms.Compose([
    transforms.ToTensor(),
    transforms.Normalize((0.5,), (0.5,))   # scales [0,1] -> [-1,1]
])
train_set_gan = torchvision.datasets.FashionMNIST(root='./data', train=True, download=True, transform=transform_gan)
train_loader_gan = DataLoader(train_set_gan, batch_size=128, shuffle=True)

g_losses, d_losses = [], []

for epoch in range(NUM_EPOCHS_GAN):
    for real_imgs, real_labels in train_loader_gan:
        batch_size = real_imgs.size(0)
        real_imgs = real_imgs.to(device)
        real_labels = real_labels.to(device)

        real_targets = torch.full((batch_size, 1), REAL_LABEL_SMOOTH, device=device)
        fake_targets = torch.zeros((batch_size, 1), device=device)

        # ---------------------
        # Train Discriminator
        # ---------------------
        d_optimizer.zero_grad()

        # Real images
        d_real_out = discriminator(real_imgs, real_labels)
        d_real_loss = criterion(d_real_out, real_targets)

        # Fake images
        z = torch.randn(batch_size, Z_DIM, device=device)
        fake_labels = torch.randint(0, NUM_CLASSES, (batch_size,), device=device)
        fake_imgs = generator(z, fake_labels)
        d_fake_out = discriminator(fake_imgs.detach(), fake_labels)
        d_fake_loss = criterion(d_fake_out, fake_targets)

        d_loss = d_real_loss + d_fake_loss
        d_loss.backward()
        d_optimizer.step()

        # ---------------------
        # Train Generator
        # ---------------------
        g_optimizer.zero_grad()

        d_out_for_g = discriminator(fake_imgs, fake_labels)
        g_loss = criterion(d_out_for_g, real_targets)   # wants D to think fakes are real

        g_loss.backward()
        g_optimizer.step()

    g_losses.append(g_loss.item())
    d_losses.append(d_loss.item())
    print(f"Epoch [{epoch+1}/{NUM_EPOCHS_GAN}] - D Loss: {d_loss.item():.4f} | G Loss: {g_loss.item():.4f}")

    # Visualize one sample per class after each epoch
    generator.eval()
    with torch.no_grad():
        z_fixed = torch.randn(10, Z_DIM, device=device)
        labels_fixed = torch.arange(0, 10, device=device)
        samples = generator(z_fixed, labels_fixed)
        samples = (samples + 1) / 2   # rescale [-1,1] -> [0,1] for display
    show_grid(samples, nrow=10, title=f"Epoch {epoch+1}: One sample per class")
    generator.train()

# Plot losses
plt.figure(figsize=(6, 4))
plt.plot(g_losses, label="Generator")
plt.plot(d_losses, label="Discriminator")
plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.title("cGAN Training Losses")
plt.legend()
plt.grid(True)
plt.show()
```

```
# ============================================
# 5.1 StyleGAN-lite: AdaIN
# ============================================

# --- A small conv autoencoder whose bottleneck is a feature MAP (not flattened) ---
class ConvEncoder(nn.Module):
    def __init__(self):
        super().__init__()
        self.net = nn.Sequential(
            nn.Conv2d(1, 32, kernel_size=3, stride=2, padding=1),   # 28x28 -> 14x14
            nn.ReLU(True),
            nn.Conv2d(32, 64, kernel_size=3, stride=2, padding=1),  # 14x14 -> 7x7
            nn.ReLU(True)
        )

    def forward(self, x):
        return self.net(x)   # shape: (B, 64, 7, 7)


class ConvDecoder(nn.Module):
    def __init__(self):
        super().__init__()
        self.net = nn.Sequential(
            nn.ConvTranspose2d(64, 32, kernel_size=4, stride=2, padding=1),  # 7x7 -> 14x14
            nn.ReLU(True),
            nn.ConvTranspose2d(32, 1, kernel_size=4, stride=2, padding=1),   # 14x14 -> 28x28
            nn.Sigmoid()
        )

    def forward(self, feat):
        return self.net(feat)


class ConvAE(nn.Module):
    def __init__(self):
        super().__init__()
        self.encoder = ConvEncoder()
        self.decoder = ConvDecoder()

    def forward(self, x):
        feat = self.encoder(x)
        return self.decoder(feat)


# Quick training so the decoder can render feature maps as real-looking images
style_ae = ConvAE().to(device)
style_optimizer = optim.Adam(style_ae.parameters(), lr=1e-3)
style_criterion = nn.MSELoss()

STYLE_AE_EPOCHS = 5
for epoch in range(STYLE_AE_EPOCHS):
    running_loss = 0.0
    for images, _ in train_loader:
        images = images.to(device)
        style_optimizer.zero_grad()
        recon = style_ae(images)
        loss = style_criterion(recon, images)
        loss.backward()
        style_optimizer.step()
        running_loss += loss.item() * images.size(0)
    print(f"[Style AE] Epoch [{epoch+1}/{STYLE_AE_EPOCHS}] - MSE: {running_loss/len(train_loader.dataset):.6f}")


# --- AdaIN function ---
def adain(content_feat, style_feat, eps=1e-5):
    """
    Adaptive Instance Normalization:
    Normalize content features to zero mean/unit std (per-channel, per-sample),
    then rescale using the style features' mean and std.
    content_feat, style_feat: (B, C, H, W)
    """
    c_mean = content_feat.mean(dim=[2, 3], keepdim=True)
    c_std = content_feat.std(dim=[2, 3], keepdim=True) + eps

    s_mean = style_feat.mean(dim=[2, 3], keepdim=True)
    s_std = style_feat.std(dim=[2, 3], keepdim=True) + eps

    normalized = (content_feat - c_mean) / c_std   # normalize content
    stylized = normalized * s_std + s_mean          # re-scale with style stats
    return stylized


# --- Pick 8 content images and 8 style images ---
style_ae.eval()
with torch.no_grad():
    data_iter = iter(test_loader)
    batch_imgs, batch_labels = next(data_iter)

    content_imgs = batch_imgs[:8].to(device)
    style_imgs = batch_imgs[8:16].to(device)   # different images for style

    content_feat = style_ae.encoder(content_imgs)   # (8, 64, 7, 7)
    style_feat = style_ae.encoder(style_imgs)       # (8, 64, 7, 7)

    mixed_feat = adain(content_feat, style_feat)
    mixed_imgs = style_ae.decoder(mixed_feat)

    # Sanity check: also decode content/style directly through the trained decoder
    content_recon = style_ae.decoder(content_feat)
    style_recon = style_ae.decoder(style_feat)
```
